```python
from collections import deque
import time


def bfs(start, graph, n):
    visited = [False] * (n + 1)
    queue = deque()

    visited[start] = True
    queue.append(start)

    traversal = []

    while queue:
        current = queue.popleft()
        traversal.append(current)

        for neighbor in graph[current]:
            if not visited[neighbor]:
                visited[neighbor] = True
                queue.append(neighbor)

    # Visit disconnected vertices
    for i in range(1, n + 1):
        if not visited[i]:
            visited[i] = True
            queue.append(i)

            while queue:
                current = queue.popleft()
                traversal.append(current)

                for neighbor in graph[current]:
                    if not visited[neighbor]:
                        visited[neighbor] = True
                        queue.append(neighbor)

    return traversal


# Input
n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

graph = [[] for _ in range(n + 1)]

print("Enter edges (u v):")

for _ in range(e):
    u, v = map(int, input().split())

    # Undirected graph
    graph[u].append(v)
    graph[v].append(u)

start = int(input("Enter starting vertex: "))

# Measure execution time
start_time = time.perf_counter()

result = bfs(start, graph, n)

end_time = time.perf_counter()

execution_time = end_time - start_time

print("\nBFS Traversal:", *result)
print("Execution Time:", execution_time, "seconds")
print("Execution Time:", execution_time * 1_000_000, "microseconds")
```

```python
import time


def dfs(current, graph, visited, traversal):
    visited[current] = True
    traversal.append(current)

    for neighbor in graph[current]:
        if not visited[neighbor]:
            dfs(neighbor, graph, visited, traversal)


def dfs_complete(start, graph, n):
    visited = [False] * (n + 1)
    traversal = []

    # Start from given vertex
    dfs(start, graph, visited, traversal)

    # Visit disconnected vertices
    for i in range(1, n + 1):
        if not visited[i]:
            dfs(i, graph, visited, traversal)

    return traversal


# Input
n = int(input("Enter number of vertices: "))
e = int(input("Enter number of edges: "))

graph = [[] for _ in range(n + 1)]

print("Enter edges (u v):")

for _ in range(e):
    u, v = map(int, input().split())

    # Undirected graph
    graph[u].append(v)
    graph[v].append(u)

start = int(input("Enter starting vertex: "))

# Measure execution time
start_time = time.perf_counter()

result = dfs_complete(start, graph, n)

end_time = time.perf_counter()

execution_time = end_time - start_time

print("\nDFS Traversal:", *result)
print("Execution Time:", execution_time, "seconds")
print("Execution Time:", execution_time * 1_000_000, "microseconds")
```
